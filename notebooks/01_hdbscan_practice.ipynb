## Mock Data UMAP Görselleştirmesi

12 hastalık grubu için mock embedding'leri 2D UMAP uzayına indirge ve scatter plot üret.

```python
import sys
import os

# src/ is one level up from notebooks/; find it regardless of CWD
_cwd = os.getcwd()
for _candidate in [
    os.path.join(_cwd, "src"),        # if CWD = project root
    os.path.join(_cwd, "..", "src"),  # if CWD = notebooks/
]:
    _candidate = os.path.abspath(_candidate)
    if os.path.isdir(_candidate) and _candidate not in sys.path:
        sys.path.insert(0, _candidate)

import umap
from mock_data import generate_mock_embeddings

embeddings, labels = generate_mock_embeddings(seed=42)

reducer = umap.UMAP(n_components=2, n_neighbors=15, min_dist=0.1, random_state=42)
umap_2d = reducer.fit_transform(embeddings)

import matplotlib.pyplot as plt
plt.figure(figsize=(10, 8))
scatter = plt.scatter(umap_2d[:, 0], umap_2d[:, 1], c=labels,
                      cmap="tab20", s=10, alpha=0.7)
plt.colorbar(scatter, label="Disease Group")
plt.title("Mock Embeddings — UMAP Projection (12 Disease Groups)")
plt.xlabel("UMAP-1")
plt.ylabel("UMAP-2")
plt.tight_layout()

# outputs/ klasörü proje root'unda; CWD'ye göre path ayarla
_outputs_dir = None
for _candidate in [
    os.path.join(_cwd, "outputs"),
    os.path.join(_cwd, "..", "outputs"),
]:
    if os.path.isdir(os.path.abspath(_candidate)):
        _outputs_dir = os.path.abspath(_candidate)
        break

if _outputs_dir:
    plt.savefig(os.path.join(_outputs_dir, "mock_umap.png"), dpi=150)
    print(f"Kaydedildi: {os.path.join(_outputs_dir, 'mock_umap.png')}")
plt.show()
```
